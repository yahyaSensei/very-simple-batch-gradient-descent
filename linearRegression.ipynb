```python
import numpy as np
import matplotlib.pyplot as plt

plt.style.use('dark_background')



x=np.arange(1,11)
y=np.arange(1,11)
print(x)
print(y)
```

```python
plt.plot(x,y,'oy')
plt.show()
```

```python
eta=0.01
epochs=1000
costValue=np.zeros(epochs)
```

```python
np.random.seed(2)

w=np.random.rand(1)[0]
b=np.random.rand(1)[0]
plt.ion() 
plt.figure(figsize=(10, 6))
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.scatter(x, y, color='yellow', label='Data Points') 
for i in range(epochs):
    y_predicated=w*x+b
    dw=(y_predicated-y)@x/len(x)
    db=np.sum(y_predicated-y)/len(x)
    w=w-eta*dw
    b=b-eta*db
    costValue[i]=0.5*np.sum((y_predicated-y)**2)/len(x)
    if(costValue[i]<0.01):
        print(f"epoch {i} cost {costValue[i]} w {w} b {b}")
        break    
    plt.scatter(x, y, color='yellow', label='Data Points') 
    
    plt.plot(x, y_predicated, color='cyan', label='Fitted Line')
    plt.grid(True, which='both', linestyle='--', linewidth=0.5)
    plt.pause(0.01)
plt.show()
```
